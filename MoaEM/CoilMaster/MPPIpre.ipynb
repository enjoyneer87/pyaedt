(3) Eddy Current 또는 AC 해석에서 계산하는 방법

AC 해석을 수행할 경우, Maxwell에서 자동으로 손실 $(P)$ 을 계산할 수 있습니다.
1. 저장된 에너지 계산
- Stored Energy ( $W$ ) 값은 Field Calculator 또는 Energy Reports 에서 확인.
- Results $\rightarrow$ Field Calculator 에서 Stored Magnetic Energy 값을 출력.
2. 손실 에너지 계산
- Results $\rightarrow$ Solution Data 에서 Eddy Current Loss 또는 Resistive Loss 값을 확인.
- $P=$ Power Loss (Watt)
3. 주파수 $(f)$ 확인
- 설정한 주파수 $(f)$ 를 확인하고, $\omega=2 \pi f$ 를 계산.
4. Q -factor 계산
- $Q=\frac{2 \pi f W}{P}$ 을 적용하여 품질 계수를 구함.

```python
def calcQualityFactor(freq, storedEnergy, lossEnergy):
    import numpy as np
    Q=2*np.pi*freq*storedEnergy/lossEnergy
    return Q
```
