```python
import xml.etree.ElementTree as ET
import svgwrite

# SVG 파일 열기
tree = ET.parse(r"C:\Users\user\Downloads\testPath.svg")
root = tree.getroot()

# SVG namespace 정의
namespace = {'svg': 'http://www.w3.org/2000/svg'}
ET.register_namespace('', 'http://www.w3.org/2000/svg')

# 원본 SVG의 크기와 뷰포트 가져오기
svg_width = root.attrib.get('width', '1200px')
svg_height = root.attrib.get('height', '700px')
svg_viewbox = root.attrib.get('viewBox', None)

# 추출된 데이터 저장
circles = []
texts = []
edges = []

# <circle> 태그 추출
for elem in root.findall(".//svg:circle", namespace):
    cx = float(elem.attrib.get('cx', 0))
    cy = float(elem.attrib.get('cy', 0))
    r = float(elem.attrib.get('r', 0))
    fill = elem.attrib.get('fill', 'none')
    stroke = elem.attrib.get('stroke', 'black')
    stroke_width = elem.attrib.get('stroke-width', '1')
    circles.append((cx, cy, r, fill, stroke, stroke_width))
circles
```

```python

# <text> 태그 추출
for elem in root.findall(".//svg:text", namespace):
    x = float(elem.attrib.get('x', 0))
    y = float(elem.attrib.get('y', 0))
    content = "".join(elem.itertext())
    fill = elem.attrib.get('fill', 'black')
    font_size = elem.attrib.get('font-size', '12px')
    texts.append((x, y, content, fill, font_size))

# <line> 태그 추출
for elem in root.findall(".//svg:line", namespace):
    x1 = float(elem.attrib.get('x1', 0))
    y1 = float(elem.attrib.get('y1', 0))
    x2 = float(elem.attrib.get('x2', 0))
    y2 = float(elem.attrib.get('y2', 0))
    stroke = elem.attrib.get('stroke', '#888')
    stroke_width = elem.attrib.get('stroke-width', '2')
    edges.append(((x1, y1), (x2, y2), stroke, stroke_width))

# 새로운 SVG 파일 생성
dwg = svgwrite.Drawing("recreated_svg_with_attributes.svg", profile='full', size=(svg_width, svg_height))

# 뷰포트 설정 (원본 SVG와 동일하게 설정)
if svg_viewbox:
    dwg.viewbox(*map(float, svg_viewbox.split()))

# 엣지(선) 추가
for (x1, y1), (x2, y2), stroke, stroke_width in edges:
    dwg.add(dwg.line(start=(x1, y1), end=(x2, y2), stroke=stroke, stroke_width=float(stroke_width)))

# 노드(원) 추가
for cx, cy, r, fill, stroke, stroke_width in circles:
    dwg.add(dwg.circle(center=(cx, cy), r=r, fill=fill, stroke=stroke, stroke_width=float(stroke_width)))

# 텍스트 추가
for x, y, content, fill, font_size in texts:
    dwg.add(dwg.text(content, insert=(x, y), fill=fill, font_size=font_size))

# SVG 파일 저장
dwg.save()

print("SVG 파일이 'recreated_svg_with_attributes.svg'로 저장되었습니다.")
```

```python
import svgwrite
import networkx as nx

# 노드와 관계 정의
nodes = [
    "Cell ECM (TB)", "Module ECM (TB)", "Pack ECM (TB)",
    "Cell ETC (FL)", "Cell CHT (FL)", "Cell CHT+ Thermal Abuse",
    "Module ETC (FL)", "Module CHT (FL)", "Module CHT+ Thermal Abuse",
    "Module LTI ROM (FL)", "Module SVD ROM (FL)", "Module ETC (TB, FL)",
    "System Tool", "Pack ETC (TB, FL)", "Pack CHT (FL)", "Pack ETC (FL)",
    "Cold Plate CHT (FL)", "Module CFD/SVD ROM (FL,TB)"
]

edges = [
    ("Cell ECM (TB)", "Cell ETC (FL)"),
    ("Cell ETC (FL)", "Cell CHT (FL)"),
    ("Cell CHT (FL)", "Cell CHT+ Thermal Abuse"),
    ("Module ECM (TB)", "Module ETC (FL)"),
    ("Module ECM (TB)", "Module ETC (TB, FL)"),
    ("Module ETC (FL)", "Module CHT (FL)"),
    ("Module CHT (FL)", "Module CHT+ Thermal Abuse"),
    ("Module CHT (FL)", "Module LTI ROM (FL)"),
    ("Module CHT (FL)", "Module SVD ROM (FL)"),
    ("Pack ECM (TB)", "Pack ETC (TB, FL)"),
    ("Pack ETC (TB, FL)", "Pack CHT (FL)"),
    ("Pack ETC (TB, FL)", "Pack ETC (FL)"),
    ("Pack ETC (TB, FL)", "Cold Plate CHT (FL)"),
    ("Pack ETC (TB, FL)", "Module CFD/SVD ROM (FL,TB)")
]

# 그래프 생성
G = nx.DiGraph()
G.add_nodes_from(nodes)
G.add_edges_from(edges)

# 노드 위치 자동 계산 (spring_layout 사용)
positions = nx.spring_layout(G, scale=700, center=(600, 350))  # SVG 크기에 맞게 조정

# SVG 생성
dwg = svgwrite.Drawing("node_descriptions_auto.svg", profile='full', size=("1200px", "700px"))

# 엣지(선) 추가
for src, tgt in edges:
    x1, y1 = positions[src]
    x2, y2 = positions[tgt]
    dwg.add(dwg.line(start=(x1, y1), end=(x2, y2), stroke="#888", stroke_width=2))

# 노드(텍스트) 추가
for node, (x, y) in positions.items():
    dwg.add(dwg.circle(center=(x, y), r=20, fill="lightblue", stroke="black", stroke_width=1))  # 노드 원
    dwg.add(dwg.text(node,
                     insert=(x, y - 25),  # 텍스트를 노드 위에 배치
                     fill="black",
                     font_size="12px",
                     text_anchor="middle"))  # 텍스트 중앙 정렬

# SVG 저장
dwg.save()

print("SVG 파일이 'node_descriptions_auto.svg'로 저장되었습니다.")
```
