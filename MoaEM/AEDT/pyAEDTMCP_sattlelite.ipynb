# Maxwell 3D 디자인 추가 - 위성 Geometry

Maxwell 3D 디자인을 추가하고 위성의 복합 geometry를 생성해보겠습니다.

```python
# Maxwell 3D 인스턴스 생성
print("\n=== Maxwell 3D 디자인 생성 ===")

# Maxwell 3D 새 디자인 생성
maxwell_3d = pyaedt.Maxwell3d(
    version=AEDT_VERSION,
    new_desktop=False,  # 기존 desktop 사용
    non_graphical=NG_MODE,
    design="Satellite_Design"
)

print(f"✓ Maxwell 3D 디자인이 생성되었습니다!")
print(f"  Project Name: {maxwell_3d.project_name}")
print(f"  Design Name: {maxwell_3d.design_name}")
print(f"  Solution Type: {maxwell_3d.solution_type}")

# 단위를 meter로 설정 (위성용)
maxwell_3d.modeler.model_units = "meter"
print(f"  Model Units: {maxwell_3d.modeler.model_units}")

# Desktop을 visible로 설정
maxwell_3d.desktop_class.visible = True
```

```python
# 위성 메인 몸체 생성
print("\n=== 위성 메인 몸체 생성 ===")

# 위성 메인 몸체 (육각형 프리즘)
main_body = maxwell_3d.modeler.create_box(
    origin=[0, 0, 0],
    sizes=[2, 1.5, 1],  # 2m x 1.5m x 1m
    name="Satellite_Main_Body"
)
main_body.material_name = "aluminum"
main_body.color = (192, 192, 192)  # 은색
print(f"메인 몸체 생성: {main_body.name}")

# 위성 안테나 디시 (실린더)
antenna_dish = maxwell_3d.modeler.create_cylinder(
    cs_axis="Z",
    position=[0, 0, 1],  # 메인 몸체 위에
    radius=0.8,
    height=0.1,
    name="Antenna_Dish"
)
antenna_dish.material_name = "copper"
antenna_dish.color = (255, 215, 0)  # 금색
print(f"안테나 디시 생성: {antenna_dish.name}")

# 안테나 피드혼 (작은 원뿔)
feed_horn = maxwell_3d.modeler.create_cone(
    cs_axis="Z",
    position=[0, 0, 1.1],
    bottom_radius=0.1,
    top_radius=0.05,
    height=0.2,
    name="Feed_Horn"
)
feed_horn.material_name = "copper"
feed_horn.color = (255, 215, 0)  # 금색
print(f"피드혼 생성: {feed_horn.name}")
```

```python
# 위성 태양전지판 생성
print("\n=== 태양전지판 및 추가 구성요소 생성 ===")

# 좌측 태양전지판
left_solar_panel = maxwell_3d.modeler.create_box(
    origin=[-3, -0.75, 0.25],
    sizes=[2, 1.5, 0.05],  # 얇은 판형태
    name="Left_Solar_Panel"
)
left_solar_panel.material_name = "silicon"
left_solar_panel.color = (0, 0, 139)  # 진한 파란색
print(f"좌측 태양전지판 생성: {left_solar_panel.name}")

# 우측 태양전지판
right_solar_panel = maxwell_3d.modeler.create_box(
    origin=[2, -0.75, 0.25],
    sizes=[2, 1.5, 0.05],
    name="Right_Solar_Panel"
)
right_solar_panel.material_name = "silicon"
right_solar_panel.color = (0, 0, 139)  # 진한 파란색
print(f"우측 태양전지판 생성: {right_solar_panel.name}")

# 태양전지판 연결 암 (좌측)
left_arm = maxwell_3d.modeler.create_cylinder(
    cs_axis="X",
    position=[-1, 0, 0.275],
    radius=0.05,
    height=1,
    name="Left_Panel_Arm"
)
left_arm.material_name = "aluminum"
left_arm.color = (128, 128, 128)  # 회색
print(f"좌측 연결암 생성: {left_arm.name}")

# 태양전지판 연결 암 (우측)
right_arm = maxwell_3d.modeler.create_cylinder(
    cs_axis="X",
    position=[1, 0, 0.275],
    radius=0.05,
    height=1,
    name="Right_Panel_Arm"
)
right_arm.material_name = "aluminum"
right_arm.color = (128, 128, 128)  # 회색
print(f"우측 연결암 생성: {right_arm.name}")

# 통신 안테나 (막대형)
comm_antenna = maxwell_3d.modeler.create_cylinder(
    cs_axis="Y",
    position=[0.5, 0.75, 0.5],
    radius=0.02,
    height=1,
    name="Communication_Antenna"
)
comm_antenna.material_name = "copper"
comm_antenna.color = (255, 0, 0)  # 빨간색
print(f"통신 안테나 생성: {comm_antenna.name}")
```

```python
# 추가 위성 구성요소
print("\n=== 추가 구성요소 및 마무리 ===")

# 추진 시스템 (작은 원뿔들)
thruster_positions = [
    [0.8, 0.6, -0.1],   # 우측 상단
    [0.8, -0.6, -0.1],  # 우측 하단
    [-0.8, 0.6, -0.1],  # 좌측 상단
    [-0.8, -0.6, -0.1]  # 좌측 하단
]

for i, pos in enumerate(thruster_positions):
    # 추진기 위치를 아래쪽으로 조정
    thruster_pos = [pos[0], pos[1], pos[2] - 0.15]
    thruster = maxwell_3d.modeler.create_cone(
        cs_axis="Z",
        position=thruster_pos,
        bottom_radius=0.08,
        top_radius=0.04,
        height=0.15,  # 양수로 수정
        name=f"Thruster_{i+1}"
    )
    if thruster:  # 생성 성공 확인
        thruster.material_name = "stainless_steel"
        thruster.color = (105, 105, 105)  # 어두운 회색
        print(f"추진기 {i+1} 생성: {thruster.name}")
    else:
        print(f"추진기 {i+1} 생성 실패")

# 센서 모듈 (작은 박스)
sensor_module = maxwell_3d.modeler.create_box(
    origin=[-0.3, -1, 0.8],
    sizes=[0.6, 0.4, 0.3],
    name="Sensor_Module"
)
sensor_module.material_name = "titanium"
sensor_module.color = (75, 0, 130)  # 인디고색
print(f"센서 모듈 생성: {sensor_module.name}")

# 생성된 모든 객체 목록 확인
all_objects = maxwell_3d.modeler.object_names
print(f"\n생성된 위성 구성요소들:")
for i, obj_name in enumerate(all_objects, 1):
    print(f"  {i}. {obj_name}")

# 뷰 설정
maxwell_3d.modeler.fit_all()
print(f"\n✓ 위성 geometry 생성 완료!")
print(f"총 {len(all_objects)}개의 구성요소가 생성되었습니다.")
print("Maxwell 3D GUI에서 위성 모델을 확인할 수 있습니다.")
```
